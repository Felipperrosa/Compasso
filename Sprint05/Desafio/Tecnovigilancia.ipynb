```python
import pandas as pd
import numpy as np


#Carregar o arquivo CSV original 

arquivo = "DADOS_ABERTOS_TECNOVIGILANCIA.csv"
df  = pd.read_csv(arquivo, sep=';', encoding='latin1')

# total_rows_initial = len(df )
# print(f"Número de linhas: {total_rows_initial}")


# 1. Filtrar dados usando dois operadores lógicos 
filtro_df = df[(df['UF_EMPRESA_NOTIFICANTE'] == "RJ") & (df['ANO_NOTIFICACAO'] <= 2022)].copy()

# 2. Duas funções de Agregação

#  2.1 Contar ocorrencia de cada 'TIPO_NOTIFICAÇÃO'

tipo = filtro_df.groupby("TIPO_NOTIFICACAO").size()
 
#  2.2 Contar notificações por estado de ocorrência 

notificações = filtro_df .groupby("UF_OCORRENCIA").size()

# print("\nContagem de ocorrências por TIPO_NOTIFICACAO:")
# print(tipo)

# print("\nMédia de notificações por UF_OCORRENCIA :")
# print(notificações)

# 3. Função Condicional  

filtro_df ["ALTO_RISCO"] = filtro_df ["CLASSE_RISCO"].apply(lambda x: "Sim" if x in ["III","IV"] else "Não")

# print("\nNova coluna 'ALTO_RISCO' adicionada:")
# print(filtro_df [["CLASSE_RISCO", "ALTO_RISCO"]].head(50))



# 5. Função data
filtro_df ['DATA_NOTIFICACAO'] = pd.to_datetime(filtro_df ['DATA_NOTIFICACAO'],errors='coerce')
filtro_df ["MES_NOTIFICACAO"] = filtro_df ["DATA_NOTIFICACAO"].dt.month

# print("\nData modificada:")
# print(filtro_df ['MES_NOTIFICACAO'].head())


#  4. Função Conversão
filtro_df ['DATA_NOTIFICACAO'] = pd.to_datetime(filtro_df ['DATA_NOTIFICACAO'],errors='coerce').dt.strftime('%d/%m/%Y')

# print("\nData modificada:")
# print(filtro_df ['DATA_NOTIFICACAO'].head())



# 6. Função string 
filtro_df = filtro_df.map(lambda x: x.replace('Tï¿½cnica', 'Técnica')
                          .replace('Mï¿½dico', 'Médico')
                          .replace('Presenï¿½a', 'Presença')
                          .replace('Nï¿½o', 'Não')
                          .replace('funï¿½ï¿½o', 'função') if isinstance(x, str) else x)

filtro_df .replace({'--': np.nan, '': np.nan}, inplace=True)

filtro_df = filtro_df.sort_values(by=["DATA_NOTIFICACAO", "ANO_NOTIFICACAO"], ascending=True)


# 8. Salvar o dataset final em um novo arquivo CSV
filtro_df .to_csv("dados_filtrados_com_modificacoes.csv", index=False, sep=';')
print("Novo dataset gerado e salvo em 'dados_filtrados_com_modificacoes.csv'")








  



```
