## Etapa 1

```python
from pyspark.sql import SparkSession
from pyspark import SparkContext, SQLContext
from pyspark.sql.functions import  expr,when, rand
from pyspark.sql import functions as F


spark = SparkSession.builder.master("local").appName("Exercicio Intro").getOrCreate()


names = "nomes_aleatorios.txt"  # Caminho relativo para subpasta


# Lendo o arquivo nomes_aleatorios.txt com Spark
df_nomes = spark.read.csv(names, header=False, inferSchema=True)

# Exibindo as primeiras 5 linhas do DataFrame
df_nomes.show(5)
```

## Etapa 2

```python
# Vendo o Schema
df_nomes.printSchema()

# Renomeando a coluna
df_nomes = df_nomes.withColumnRenamed("_c0","Nomes")

df_nomes.show(10)
```

## Etapa 3

```python
# A função rand() gera um número aleatório entre 0 e 1 para cada linha do data frame 
df_nomes = df_nomes.withColumn(
    "Escolaridade",
    # Se o número gerado for menor que 0,33 a escolaridade e fundamental assim como se o número for menor que 0,66 a escolaridade e medio
    when(rand() < 0.33, "Fundamental")
    .when(rand() < 0.66, "Medio")
    .otherwise("Superior")
)

df_nomes.show(10)
```

## Etapa 4


```python
paises = [ "Argentina", "Bolívia", "Brasil", "Chile", "Colômbia", 
"Equador", "Guiana", "Paraguai", "Peru", "Suriname", 
"Uruguai", "Venezuela", "Guiana Francesa"]

# map(lambda x: f'\"{x}\"', paises) transforma cada país da lista paises em um formato válido para a expressão SQL 
df_nomes = df_nomes.withColumn(
    "Pais",
    expr(f"array({', '.join(map(lambda x: f'\"{x}\"', paises))})[int(rand() * {len(paises)})]"
))

df_nomes.show(10)
```

## Etapa 5

```python

df_nomes = df_nomes.withColumn(
    "Ano_Nascimento",
    expr("1945 + int(rand() * 66)")  # Gera um ano de nascimento aleatório entre 1945 e 2010
)


df_nomes.show(10)
```

# Etapa 6

```python
# Filtra no Ano_Nascimento for maior ou igual a 2001
df_select = df_nomes.filter(df_nomes["Ano_Nascimento"] >= 2001)

df_select.show(10)
```

# Etapa 7

```python
df_nomes.createOrReplaceTempView("pessoas")
# Filtra no Ano_Nascimento for maior ou igual a 2001 porem em sql 
df_select = spark.sql("SELECT * FROM pessoas WHERE  Ano_Nascimento >= 2001").show()

```

# Etapa 8

```python
# Filtra o Ano_Nascimento entre 1980 e 1994
df_millennials = df_nomes.filter(F.col("Ano_Nascimento").between(1980, 1994))

df_millennials.show(10)
```

# Etapa 9

```python
df_nomes.createOrReplaceTempView("pessoas")
# Filtra o Ano_Nascimento entre 1980 e 1994 sql
df_select = spark.sql("SELECT * FROM millennials WHERE  Ano_Nascimento BETWEEN 1980 and 1994").show()
```

# Etapa 10

```python
df_nomes.createOrReplaceTempView("pessoas")

# Executa minha consulta sql 
df_select = spark.sql(
                      """
    SELECT 
        Pais,
        CASE
            WHEN Ano_Nascimento BETWEEN 1944 AND 1964 THEN 'Baby Boomers'
            WHEN Ano_Nascimento BETWEEN 1965 AND 1979 THEN 'Geração X'
            WHEN Ano_Nascimento BETWEEN 1980 AND 1994 THEN 'Millennials (Geração Y)'
            WHEN Ano_Nascimento BETWEEN 1995 AND 2015 THEN 'Geração Z'
        END AS Geracao,
        COUNT(*) AS Quantidade
    FROM pessoas
    GROUP BY Pais, Geracao
    ORDER BY Pais, Geracao, Quantidade
"""
).show()
```
